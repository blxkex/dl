## Math functions Needed.
---
predefining some math functions that i would need.

```python
def sigmoid(bruh: np.ndarray):
    return 1 / (1 + np.exp(-bruh))

def sigmoid_derivative(bruh: np.ndarray):
    return sigmoid(bruh) * (1 - sigmoid(bruh))
```

## Preprocessing

```python
x = data.iloc[:, :-1]
y_s = data.iloc[:, -1]

# converting the values from string (positive or negative) to (1 or 0)
y = []
for i in y_s:
    y.append(1 if i == 'positive' else 0)

# converting shit to numpy arrays.
x = np.array(x, dtype=np.float64)
y = np.array(y, dtype=np.float64)

sc = StandardScaler()
x = sc.fit_transform(x)
```

## Creating the Network

I'm planning on creating a network with 7 input neurons and 3 hidden layers (5, 4, 3) respectively. and obviously 1 output neuron.

```python
class Network:
    def __init__(self, inputs: np.ndarray, correct_outputs : np.ndarray):
        self.inputs = inputs
        self.correct_outputs = correct_outputs
        self.weights_1 = np.random.rand(7, 5)
        self.hidden_layer_1 = None
        self.weights_2 = np.random.rand(5, 4)
        self.hidden_layer_2 = None
        self.weights_3 = np.random.rand(4, 3)
        self.hidden_layer_3 = None
        self.weights_4 = np.random.rand(3, 1)
        self.predictions = None

        self.learning_rate = 0.01


    def testing(self):
        print(self.weights_3)

    def forward(self, idx):
        self.z1 = np.dot(self.inputs[idx], self.weights_1)
        self.hidden_layer_1 = sigmoid(self.z1)
        self.z2 = np.dot(self.hidden_layer_1, self.weights_2)
        self.hidden_layer_2 = sigmoid(self.z2)
        self.z3 = np.dot(self.hidden_layer_2, self.weights_3)
        self.hidden_layer_3 = sigmoid(self.z3)
        self.z4 = np.dot(self.hidden_layer_3, self.weights_4)
        self.predictions = sigmoid(self.z4)

        self.predictions_array.append(self.predictions)

    def backward(self, idx):
        # delta at output layer
        loss_vector_4 = (self.predictions - self.correct_outputs[idx]) * sigmoid_derivative(self.z4)
        gradients_4 = np.outer(self.hidden_layer_3, loss_vector_4)

        # hidden layer 3
        loss_vector_3 = (loss_vector_4 @ self.weights_4.T) * sigmoid_derivative(self.z3)
        gradients_3 = np.outer(self.hidden_layer_2, loss_vector_3)

        # hidden layer 2
        loss_vector_2 = (loss_vector_3 @ self.weights_3.T) * sigmoid_derivative(self.z2)
        gradients_2 = np.outer(self.hidden_layer_1, loss_vector_2)

        # hidden layer 1
        loss_vector_1 = (loss_vector_2 @ self.weights_2.T) * sigmoid_derivative(self.z1)
        gradients_1 = np.outer(self.inputs[idx], loss_vector_1)

        # update
        self.weights_4 -= self.learning_rate * gradients_4
        self.weights_3 -= self.learning_rate * gradients_3
        self.weights_2 -= self.learning_rate * gradients_2
        self.weights_1 -= self.learning_rate * gradients_1


        # training process per epoch.
    def train(self, epochs):
        for epoch in range(epochs):
            self.predictions_array = []

            for idx in range(len(self.inputs)):
                self.forward(idx)
                self.backward(idx)

            self.predictions_array = np.array(self.predictions_array).reshape(-1, 1)

            # compute loss (MSE)
            loss = np.mean((self.predictions_array - self.correct_outputs) ** 2)

            # compute accuracy (binary classification threshold)
            preds_binary = (self.predictions_array > 0.5).astype(int).reshape(-1, 1)
            labels = self.correct_outputs.reshape(-1, 1).astype(int)
            acc = np.mean(preds_binary == labels)

            print(f"epoch {epoch+1} completed. Loss = {loss:.4f}, Accuracy = {acc:.4f}")

```

```python
my_network = Network(x, y)
my_network.train(100)
```
